```python
from mallows import TopKMallows
import numpy as np
```

```python
algo_beta = 0.5
human_beta = 0.5
p_penalty = 0.321
algo_k = 2
universe = [0, 1, 2, 3]
utility_arr = np.array([0, 1, 0, 0])

human = TopKMallows(center=[1, 2, 3], k=3, beta=human_beta, p=p_penalty)

algo_1 = TopKMallows(center=[1, 2, 3], k=algo_k, beta=algo_beta, p=p_penalty)

human.collab_utility(algo_1, universe, utility_arr)
```

```python
algo_2 = TopKMallows(center=[1, 3, 2], k=algo_k, beta=algo_beta, p=p_penalty)

human.collab_utility(algo_2, universe, utility_arr)
```

```python
algo_3 = TopKMallows(center=[3, 1, 2], k=algo_k, beta=algo_beta, p=p_penalty)

human.collab_utility(algo_3, universe, utility_arr)
```

```python
algo_4 = TopKMallows(center=[3, 2, 1], k=algo_k, beta=algo_beta, p=p_penalty)

human.collab_utility(algo_4, universe, utility_arr)
```
